<h1>FOR NTU60</h1>

```python
import pickle

import numpy as np
from tqdm import tqdm
npz_data = np.load('/home/data2/sabdrakhim/data/ntu/NTU60_CV.npz')
label = np.where(npz_data['y_test'] > 0)[1]
r1 = open('work_test/ntu/xview_joint_best_acc.pkl', 'rb')
r1 = list(pickle.load(r1).items())
r2 = open('work_test/ntu/xview_bone_best_acc.pkl', 'rb')
r2 = list(pickle.load(r2).items())
'''r3 = open('work_test/ntu/xview_vel_best_acc.pkl', 'rb')
r3 = list(pickle.load(r3).items())
r4 = open('work_test/ntu/xview_bonevel_best_acc.pkl', 'rb')
r4 = list(pickle.load(r4).items())'''

alpha = [0.5, 0.5, 1, 1]
right_num = total_num = right_num_5 = 0

for i in tqdm(range(len(label))):
    l = label[i]
    _, r11 = r1[i]
    _, r22 = r2[i]
    r = r11 * alpha[0] + r22 * alpha[1] #* alpha
    rank_5 = r.argsort()[-5:]
    right_num_5 += int(int(l) in rank_5)
    r = np.argmax(r)
    right_num += int(r == int(l))
    total_num += 1
acc = right_num / total_num
acc5 = right_num_5 / total_num
print('top1: ', acc)
print('top5: ', acc5)
```
